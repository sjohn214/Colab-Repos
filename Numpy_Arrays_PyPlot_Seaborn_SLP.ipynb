#### Plotting a function y = 4*x+15

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange (2, 30)
y = x + 15

plt.title ("Plotting a Function")

plt.xlabel ("x-axis")
plt.ylabel ("y-axis")

plt.plot (x, y, color = "blue")

plt.show ()
```
