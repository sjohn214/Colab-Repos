## Convolutional Neural Network Example

```python
import tensorflow as tf
import numpy as np
from tensorflow import keras
```

```python
# mnist dataset
mnist = keras.datasets.mnist
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# training scope
EPOCHS = 5
BATCH_SIZE = 128
VERBOSE = 1
VALIDATION_SPLIT = 0.90
NB_CLASSES = 10
```

```python
# NN Model data and reshape
train_images = train_images.reshape(60000, 28, 28, 1)
test_images = test_images.reshape(10000, 28, 28, 1)

# Normalization
train_images = train_images/255.0
test_images = test_images/255.0

# float32 data type
train_images.astype('float32')
test_images.astype('float32')

#ONE-HOT Encoded format
train_labels = keras.utils.to_categorical(train_labels, NB_CLASSES)
test_labels = keras.utils.to_categorical(test_labels, NB_CLASSES)
```

## Model defining and building

```python
from keras import layers, models
cnn_model = models.Sequential()

cnn_model.add(layers.Convolution2D(20, (5, 5), activation='relu', input_shape=(28, 28, 1)))

cnn_model.add(layers.MaxPooling2D(pool_size=(2, 2), strides=(2, 2)))

# MaxPool
cnn_model.add(layers.Convolution2D(50, (5, 5), activation='relu'))
# MaxPool
cnn_model.add(layers.MaxPooling2D(pool_size=(2, 2), strides=(2, 2)))
```

```python
# Flatten
cnn_model.add(layers.Flatten())

# Dense w/500 neurons
cnn_model.add(layers.Dense(500, activation='relu'))

# Softmax
cnn_model.add(layers.Dense(NB_CLASSES, activation='softmax'))
```

```python
# Compile
cnn_model.compile(loss='categorical_crossentropy',
                  optimizer='Adam',
                  metrics=['accuracy'])
# Summary
cnn_model.summary()
```

```python
# Model fit
history = cnn_model.fit(train_images, train_labels,
                        epochs=EPOCHS,
                        batch_size=BATCH_SIZE,
                        verbose=VERBOSE,
                        validation_split=VALIDATION_SPLIT)
```

```python
# Score evaluation
evaluation = cnn_model.evaluate(test_images, test_labels, verbose=VERBOSE)

print("\nTest Evaluation:", evaluation[0])
print('Test Accuracy:', evaluation[1])
```
