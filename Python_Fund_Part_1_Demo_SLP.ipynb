#**Section II)** *Collections - Lists, Tuples, Sets and Loops, and Slicing*

A list is a collection of items stored in a variable. There are no restrictions as to what can be stored. Typically, they should be stored by relatable items such as number of fruits in a basket, or in inventory.

```python
#List of student in a class
```

```python
student_list = ["John", "Amanda", "Kathy", "Indu", "Jose", "Mariam"]
```

```python
print(student_list) # print the list and its content
```

Accessing only one item from a list using the index

```python
first_student = student_list[0] # The number you see in the [0] is called an index and each element has a unique index in the list
```

```python
print(first_student)
```

```python
third_student = student_list[2] # index in a list begins with 0 so in this case it it 0.1.2.3.4.5
```

```python
print(third_student)
```

```python
fifth_student = student_list[4]
```

```python
print(fifth_student)
```

Using for...in loop to go through the student_list
