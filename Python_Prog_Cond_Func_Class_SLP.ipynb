## Python Conditional Statements

### We are using if else statements to represent logic and conditions.

#### Buy apples or oranges based on the lowest price.

```python
# Use if statement to decide between buying apples or oranges

price_of_apples = 2
price_of_oranges = 1.5

if price_of_apples < price_of_oranges:
     print("Buy apples")
else:
     print("Buy oranges")
```

#### Buy a sofa or a car based on available budget

```python
# Buy sofa or car based on available budget

price_of_sofa = 200
price_of_car = 500

budget = 300

if price_of_sofa < budget:
     print("Buy sofa")
else:
     print("Buy car")
```

#### Use ternary operators for conditional statements

```python
is_student = False
can_drive = True if is_student else False

print(can_drive)
```

```python
# Another example

is_graduate_student = True

can_apply_for_scholarship = True if is_graduate_student else False
print("Can apply for Scholarship:", can_apply_for_scholarship)
```

#### Now a function that takes two arguments

```python
# Create a function that two arguments apples and grapes to calculate total price to be paid by customer

def calculate_price(apples, grapes):
    price_of_apples = 2.65
    price_of_grapes = 1.98
    total_price = apples * price_of_apples + grapes * price_of_grapes
    return total_price

price_to_be_paid = calculate_price(30, 25)
print("Total price to be paid:", price_to_be_paid, "dollars")
```

#### Function using global variables

```python
# Create a function that has two global variables as an argument aids as price of a suv

price_of_car = 30000
price_of_suv = 50000

def calculate_price(a, b):
    return a + b

price_to_be_paid = calculate_price(price_of_car, price_of_suv)

print("Total price to be paid:", price_to_be_paid,  "dollars")
```
